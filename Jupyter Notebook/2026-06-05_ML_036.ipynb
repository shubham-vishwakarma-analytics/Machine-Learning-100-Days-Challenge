# Day 36 — Polynomial Regression

In Project 5, we assumed a linear relationship:

Area ↑  → Price ↑ (straight line)

But in real life, many relationships are curved, not straight.

Example:

Study Hours → Marks

The increase may not always be linear.

That's where Polynomial Regression helps.

### Problem Statement

Predict Salary based on Years of Experience

```python
# Step 1: Create Dataset

import pandas as pd

data = {
    "Experience": [1,2,3,4,5,6,7,8],
    "Salary": [20,25,35,50,70,95,125,160]
}

df = pd.DataFrame(data)
```

```python
# Step 2: Features and Target

X = df[['Experience']]
y = df['Salary']
```

```python
# Step 3: Create Polynomial Features

from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree = 2)
X_poly = poly.fit_transform(X)
```

```python
# Step 4: Train Model

from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_poly, y)
```

```python
 # Step 5: Predict

prediction = model.predict(poly.transform([[9]]))

print(prediction)
```

```python
# Step 6: Calculate MAE

from sklearn.metrics import mean_absolute_error

y_pred = model.predict(X_poly)

mae = mean_absolute_error(y, y_pred)

print("MAE:", mae)
```

```python
# Step 7: Visualize

import matplotlib.pyplot as plt

plt.scatter(X, y, color = 'blue')
plt.plot(X, model.predict(X_poly), color = 'red')

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Polynomial Regression")

plt.show()
```
